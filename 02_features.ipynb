```python
from pathlib import Path
import pandas as pd
import sys
from pathlib import Path


file_path = "data/raw/diabetic_data.csv"


df = pd.read_csv(file_path, dtype=str, keep_default_na=False, na_values=["?"])

print(df.head())
```

```python
from IPython.display import display, HTML

def display_scrollable(df, max_height=500):
    html = f"""
    <div style="max-height:{max_height}px; overflow:auto; width:100%;">
        {df.to_html(index=False)}
    </div>
    """
    display(HTML(html))

# Usage
display_scrollable(df.head(10))
```

```python
NUMERIC_COLS = ["time_in_hospital", "num_lab_procedures", "num_procedures",
                "num_medications", "number_outpatient", "number_emergency",
                "number_inpatient", "number_diagnoses"]
CODE_COLS = ["admission_type_id", "discharge_disposition_id", "admission_source_id",
             "encounter_id", "patient_nbr"]

d = df.copy()   # keep df untouched so you can compare

for c in NUMERIC_COLS:
    d[c] = pd.to_numeric(d[c], errors='coerce')

for c in CODE_COLS:
    d[c] = pd.to_numeric(d[c], errors='coerce').astype("Int64")
```

```python
cols = NUMERIC_COLS + CODE_COLS

print(d[cols].dtypes)
```

```python
cols = NUMERIC_COLS + CODE_COLS

# 1. Types changed: expect int64/float64 and Int64, no 'object'
print(d[cols].dtypes)

# 2. Nothing silently became missing
#    df had no '?' in these columns (QC-003), so every NaN here was created by coerce
print("NaN created by conversion:", int(d[cols].isna().sum().sum()))   # expect 0

# 3. Values unchanged, only the type: compare as text
same = (d[cols].astype(str) == df[cols].astype(str)).all()
print("Values identical to original:", same.all())                     # expect True

# 4. Sanity: ranges look plausible
print(d[NUMERIC_COLS].describe().T[["min", "max"]])
```

```python
d["number_emergency"].quantile([0.5, 0.9, 0.99, 1.0])
```

```python
EXPIRED_OR_HOSPICE = [11, 13, 14, 19, 20, 21]
s0 = d

# 1. Mask: True where the discharge code is expired or hospice
excluded = d["discharge_disposition_id"].isin(EXPIRED_OR_HOSPICE)

# 2. Keep the rows that are NOT excluded
s1 = d[~excluded]

print(f"s0: {len(s0):,}   s1: {len(s1):,}   removed: {len(s0) - len(s1):,}")
```

```python
# a) The arithmetic adds up
print("Removed = mask count:", len(s0) - len(s1) == excluded.sum())             # True

# b) No excluded codes survived
print("Excluded codes left in s1:", s1["discharge_disposition_id"].isin(EXPIRED_OR_HOSPICE).sum())  # 0

# c) Which codes were removed, and how many of each
print(s0.loc[excluded, "discharge_disposition_id"].value_counts().sort_index())

# d) QC-001 contradiction check: can a patient who died be readmitted?
died = s0["discharge_disposition_id"].isin([11, 19, 20, 21])
hosp = s0["discharge_disposition_id"].isin([13, 14])
print("\nExpired:", died.sum()); print(s0.loc[died, "readmitted"].value_counts())
print("\nHospice:", hosp.sum()); print(s0.loc[hosp, "readmitted"].value_counts())
```

```python
s2 = s1[s1["gender"].isin(["Male", "Female"])]
print(f"s1: {len(s1):,}   s2: {len(s2):,}   removed: {len(s1) - len(s2):,}")
print(s1.loc[~s1["gender"].isin(["Male", "Female"]), "gender"].value_counts())
```

```python
s3 = s2.sort_values("encounter_id").drop_duplicates("patient_nbr", keep="first")
```

```python
# a) One row per patient
print("One row per patient:", s3["patient_nbr"].is_unique)                     # True

# b) Row count = number of distinct patients in s2
print("Rows = patients in s2:", len(s3) == s2["patient_nbr"].nunique())        # True

# c) Each kept row really is that patient's earliest encounter
earliest = s2.groupby("patient_nbr")["encounter_id"].min()
kept = s3.set_index("patient_nbr")["encounter_id"]
print("Kept = earliest:", kept.sort_index().equals(earliest.sort_index()))     # True

# d) Readmission rate before and after
print(f"30-day readmission  s2: {100*(s2['readmitted']=='<30').mean():.1f}%   "
      f"s3: {100*(s3['readmitted']=='<30').mean():.1f}%")
```

```python
a=s3.copy()
a["readmit_30"] = (a["readmitted"] == "<30").astype("Int64")
```

```python
# 1. Only 0 and 1
print(a["readmit_30"].value_counts())

# 2. Rate matches the cohort (9.0%)
print(f"Rate: {100 * a['readmit_30'].mean():.1f}%")

# 3. Every original value mapped correctly
print(pd.crosstab(a["readmitted"], a["readmit_30"]))
```

```python
AGE_MAP = {
    "[0-10)": "<30", "[10-20)": "<30", "[20-30)": "<30",
    "[30-40)": "30-60", "[40-50)": "30-60", "[50-60)": "30-60",
    "[60-70)": ">=60", "[70-80)": ">=60", "[80-90)": ">=60", "[90-100)": ">=60"
}
a["age_group"] = a["age"].map(AGE_MAP)
```

```python
print("Unmapped:", a["age_group"].isna().sum())          # 0
print(pd.crosstab(a["age"], a["age_group"]))
print(a.groupby("age_group")["readmit_30"].mean().mul(100).round(1))
```

```python
a["age_group"] = pd.Categorical(a["age_group"], categories=["<30", "30-60", ">=60"], ordered=True)
a.groupby("age_group", observed=True)["readmit_30"].mean().mul(100).round(1)
```

```python
a["race_grp"] = (a["race"]
                 .replace({  "Hispanic": "Other", "Asian": "Other", "Other": "Other"
                 })
                 .fillna( "Missing"
                 ))
```

```python
print(a["race_grp"].value_counts(dropna=False))
print(a.groupby("race_grp")["readmit_30"].mean().mul(100).round(1))
```
