```python
NUMERIC_COLS = ["time_in_hospital", "num_lab_procedures", "num_procedures",
                "num_medications", "number_outpatient", "number_emergency",
                "number_inpatient", "number_diagnoses"]
CODE_COLS = ["admission_type_id", "discharge_disposition_id", "admission_source_id",
             "encounter_id", "patient_nbr"]

d = df.copy()   # keep df untouched so you can compare

for c in NUMERIC_COLS:
    d[c] = pd.to_numeric(d[c], errors='coerce')

for c in CODE_COLS:
    d[c] = pd.to_numeric(d[c], errors='coerce').astype("Int64")
```

```python
EXPIRED_OR_HOSPICE = [11, 13, 14, 19, 20, 21]
s0 = d

# 1. Mask: True where the discharge code is expired or hospice
excluded = d["discharge_disposition_id"].isin(EXPIRED_OR_HOSPICE)

# 2. Keep the rows that are NOT excluded
s1 = d[~excluded]

print(f"s0: {len(s0):,}   s1: {len(s1):,}   removed: {len(s0) - len(s1):,}")
```

```python
# a) The arithmetic adds up
print("Removed = mask count:", len(s0) - len(s1) == excluded.sum())             # True

# b) No excluded codes survived
print("Excluded codes left in s1:", s1["discharge_disposition_id"].isin(EXPIRED_OR_HOSPICE).sum())  # 0

# c) Which codes were removed, and how many of each
print(s0.loc[excluded, "discharge_disposition_id"].value_counts().sort_index())

# d) QC-001 contradiction check: can a patient who died be readmitted?
died = s0["discharge_disposition_id"].isin([11, 19, 20, 21])
hosp = s0["discharge_disposition_id"].isin([13, 14])
print("\nExpired:", died.sum()); print(s0.loc[died, "readmitted"].value_counts())
print("\nHospice:", hosp.sum()); print(s0.loc[hosp, "readmitted"].value_counts())
```

```python
import numpy as np

high = a["A1Cresult"].isin([">7", ">8"])
changed = a["change"] == "Ch"

conditions = [
    a["A1Cresult"] == "None",      # test not performed
    a["A1Cresult"] == "Norm",      # tested, normal result
    high & changed,                # tested, high, medication changed
    high & ~changed,               # tested, high, medication not changed
]
labels = ["Not measured", "Normal", "High, med changed", "High, med not changed"]

a["hba1c_grp"] = np.select(conditions, labels, default="Unexpected")
a["hba1c_grp"] = pd.Categorical(a["hba1c_grp"], categories=labels + ["Unexpected"])
```

```python
# 1. Every row assigned: "Unexpected" should be 0
print(a["hba1c_grp"].value_counts(dropna=False))

# 2. Each A1Cresult x change combination lands in exactly one group
print(pd.crosstab([a["A1Cresult"], a["change"]], a["hba1c_grp"]))

# 3. Unadjusted readmission rate by HbA1c group
print(a.groupby("hba1c_grp", observed=True)["readmit_30"]
       .agg(n="size", rate="mean")
       .assign(rate=lambda t: (100 * t["rate"]).round(1)))
```

```python
def icd9_group(code):
    """Map one ICD-9 code (text) to a primary diagnosis group (Strack et al.)."""
    if pd.isna(code):
        return "Other"                         # missing diag_1 (QC-009)
    code = str(code).strip()
    if code[0] in ("V", "E"):                  # supplementary / external-cause codes
        return "Other"
    x = float(code)
    whole = int(x)                             # 250.83 -> 250
    if whole == 250:
        return "Diabetes"
    if 390 <= x < 460 or whole == 785:
        return "Circulatory"
    if 460 <= x < 520 or whole == 786:
        return "Respiratory"
    if 520 <= x < 580 or whole == 787:
        return "Digestive"
    if 800 <= x < 1000:
        return "Injury"
    if 710 <= x < 740:
        return "Musculoskeletal"
    if 580 <= x < 630 or whole == 788:
        return "Genitourinary"
    if 140 <= x < 240:
        return "Neoplasms"
    return "Other"
```

```python
tests = {
    "428": "Circulatory",     # heart failure
    "414": "Circulatory",     # coronary disease
    "785": "Circulatory",     # special single code
    "786": "Respiratory",
    "250.83": "Diabetes",
    "250": "Diabetes",
    "574": "Digestive",
    "820": "Injury",          # hip fracture
    "715": "Musculoskeletal",
    "599": "Genitourinary",
    "162": "Neoplasms",       # lung cancer
    "38": "Other",            # septicemia (infectious, not in any group)
    "V57": "Other",
    "E909": "Other",
    None: "Other",
}
for code, expected in tests.items():
    got = icd9_group(code)
    print(f"{str(code):8} -> {got:16} {'OK' if got == expected else 'FAIL (expected ' + expected + ')'}")
    
```

```python
a["diag1_group"] = a["diag_1"].map(icd9_group)

# 1. Distribution and readmission rate by group
summary = (a.groupby("diag1_group")["readmit_30"]
             .agg(n="size", rate="mean")
             .assign(pct_of_cohort=lambda t: (100 * t["n"] / t["n"].sum()).round(1),
                     rate=lambda t: (100 * t["rate"]).round(1))
             .sort_values("n", ascending=False))
print(summary)

# 2. Audit section 9: how many V, E and missing codes went to Other
d1 = a["diag_1"]
print("\nMissing:", d1.isna().sum(),
      "| V codes:", d1.str.startswith("V", na=False).sum(),
      "| E codes:", d1.str.startswith("E", na=False).sum())

# 3. Sanity check: the most common codes in "Other", to see what's in there
print("\nTop codes in Other:")
print(a.loc[a["diag1_group"] == "Other", "diag_1"].value_counts().head(10))
```

```python
# --- Admission source ---
SOURCE_MAP = {7: "Emergency room",
              1: "Referral", 2: "Referral", 3: "Referral",
              9: "Unknown", 15: "Unknown", 17: "Unknown", 20: "Unknown", 21: "Unknown"}
a["admission_source_grp"] = a["admission_source_id"].map(SOURCE_MAP).fillna("Other")

# --- Discharge disposition ---
DISCHARGE_MAP = {1: "Home", 18: "Unknown", 25: "Unknown", 26: "Unknown"}
a["discharge_grp"] = a["discharge_disposition_id"].map(DISCHARGE_MAP).fillna("Other")

# --- Admission type ---
TYPE_MAP = {1: "Emergency", 2: "Urgent", 3: "Elective",
            5: "Unknown", 6: "Unknown", 8: "Unknown"}
a["admission_type_grp"] = a["admission_type_id"].map(TYPE_MAP).fillna("Other")

# --- Medical specialty ---
spec = a["medical_specialty"]
a["specialty_grp"] = np.select(
    [spec.isna(),
     spec == "InternalMedicine",
     spec == "Cardiology",
     spec == "Family/GeneralPractice",
     spec.str.startswith("Surgery", na=False)],
    ["Missing", "Internal medicine", "Cardiology", "Family/general practice", "Surgery"],
    default="Other",
)
```

```python
# 1. Each code lands in exactly one group (one non-zero cell per row)
print(pd.crosstab(a["admission_source_id"], a["admission_source_grp"]))
print(pd.crosstab(a["discharge_disposition_id"], a["discharge_grp"]))
print(pd.crosstab(a["admission_type_id"], a["admission_type_grp"]))

# 2. Specialty: what went into Surgery and Other
print(a.loc[a["specialty_grp"] == "Surgery", "medical_specialty"].value_counts().head(8))
print(a.loc[a["specialty_grp"] == "Other", "medical_specialty"].value_counts().head(8))

# 3. Group sizes and readmission rate for all four variables
for col in ["admission_source_grp", "discharge_grp", "admission_type_grp", "specialty_grp"]:
    print(f"\n{col}")
    print(a.groupby(col)["readmit_30"].agg(n="size", rate="mean")
           .assign(pct=lambda t: (100 * t["n"] / len(a)).round(1),
                   rate=lambda t: (100 * t["rate"]).round(1))
           .sort_values("n", ascending=False))
```

```python
MED_COLS = ["metformin", "repaglinide", "nateglinide", "chlorpropamide", "glimepiride",
            "acetohexamide", "glipizide", "glyburide", "tolbutamide", "pioglitazone",
            "rosiglitazone", "acarbose", "miglitol", "troglitazone", "tolazamide",
            "examide", "citoglipton", "insulin", "glyburide-metformin",
            "glipizide-metformin", "glimepiride-pioglitazone",
            "metformin-rosiglitazone", "metformin-pioglitazone"]

# 1. Only No / Steady / Up / Down anywhere?
values = pd.unique(a[MED_COLS].values.ravel())
print("Values found:", values)

# 2. Usage per drug: counts of each value + % of patients on it
usage = a[MED_COLS].apply(lambda s: s.value_counts()).T.fillna(0).astype(int)
usage["pct_used"] = (100 * (1 - usage["No"] / len(a))).round(2)
print(usage.sort_values("pct_used", ascending=False))
```

```python
meds = a[MED_COLS]

a["n_meds_used"]      = meds.ne("No").sum(axis=1)                  # drugs prescribed
a["n_meds_changed"]   = meds.isin(["Up", "Down"]).sum(axis=1)      # drugs with a dose change
a["insulin_used"]     = (a["insulin"] != "No").astype(int)
a["insulin_adjusted"] = a["insulin"].isin(["Up", "Down"]).astype(int)
a["any_med_change"]   = (a["change"] == "Ch").astype(int)          # dataset's own summary flag
a["on_diabetes_med"]  = (a["diabetesMed"] == "Yes").astype(int)    # dataset's own summary flag
```

```python
# 1. Is "on diabetes medication" consistent with prescribing at least one drug?
print(pd.crosstab(a["on_diabetes_med"], a["n_meds_used"] > 0,
                  rownames=["diabetesMed = Yes"], colnames=["any drug prescribed"]))

# 2. Is "medication changed" consistent with at least one Up/Down?
print(pd.crosstab(a["any_med_change"], a["n_meds_changed"] > 0,
                  rownames=["change = Ch"], colnames=["any Up/Down"]))

# 3. Distributions and readmission rate
for col in ["n_meds_used", "n_meds_changed", "insulin_adjusted", "any_med_change"]:
    print(f"\n{col}")
    print(a.groupby(col)["readmit_30"].agg(n="size", rate="mean")
           .assign(rate=lambda t: (100 * t["rate"]).round(1)))
```

```python
a["n_meds_used_grp"] = a["n_meds_used"].clip(upper=3)        # 0, 1, 2, 3 (= 3 or more)
a["n_meds_changed_grp"] = a["n_meds_changed"].clip(upper=2)  # 0, 1, 2 (= 2 or more)
print(a["n_meds_used_grp"].value_counts().sort_index())      # 0, 1, 2, 3 only
print(a["n_meds_changed_grp"].value_counts().sort_index())   # 0, 1, 2 only
```

```python
NUMERIC_COLS = ["time_in_hospital", "num_lab_procedures", "num_procedures",
                "num_medications", "number_outpatient", "number_emergency",
                "number_inpatient", "number_diagnoses"]

a["sex"] = a["gender"]          # clearer name for tables

KEEP = (
    ["patient_nbr", "encounter_id"]                                   # identifiers
    + ["readmit_30"]                                                  # outcome
    + ["hba1c_grp"]                                                   # primary exposure
    + ["age_group", "sex", "race_grp"]                                # demographics
    + ["admission_type_grp", "admission_source_grp",
       "discharge_grp", "specialty_grp", "diag1_group"]               # admission & diagnosis
    + ["n_meds_used", "n_meds_used_grp", "n_meds_changed", "n_meds_changed_grp",
       "insulin_used", "insulin_adjusted", "any_med_change", "on_diabetes_med"]  # medications
    + NUMERIC_COLS                                                    # counts
)
analysis = a[KEEP].copy()
print(analysis.shape)
```

```python
CATEGORICAL = ["hba1c_grp", "age_group", "sex", "race_grp", "admission_type_grp",
               "admission_source_grp", "discharge_grp", "specialty_grp", "diag1_group"]

checks = {
    "Rows = 69,987":              len(analysis) == 69987,
    "One row per patient":        analysis["patient_nbr"].is_unique,
    "No missing values anywhere": analysis.isna().sum().sum() == 0,
    "Outcome is 0/1 only":        set(analysis["readmit_30"].unique()) <= {0, 1},
    "Readmission rate = 9.0%":    round(100 * analysis["readmit_30"].mean(), 1) == 9.0,
    "No 'Unexpected' HbA1c":      (analysis["hba1c_grp"] != "Unexpected").all(),
    "Every category has 2+ levels": all(analysis[c].nunique() >= 2 for c in CATEGORICAL),
}
for name, passed in checks.items():
    print(f"{'PASS' if passed else 'FAIL'}  {name}")

# Sparse levels: any category level with fewer than 100 patients gives unstable estimates
print("\nLevels with < 100 patients:")
for c in CATEGORICAL:
    small = analysis[c].value_counts()[lambda s: s < 100]
    if len(small):
        print(f"  {c}: {small.to_dict()}")
```
