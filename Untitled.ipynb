```python
# Is missingness related to 30-day readmission?
for col in ["payer_code", "medical_specialty", "race"]:
    rate = (df.assign(is_missing=df[col].isna(),
                      readmit_30=(df["readmitted"] == "<30"))
              .groupby("is_missing")["readmit_30"]
              .mean()
              .mul(100)
              .round(1))
    print(f"{col}: readmit <30 %  present={rate[False]}  missing={rate[True]}")
```
