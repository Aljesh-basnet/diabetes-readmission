```python
from pathlib import Path
import pandas as pd

file_path = "data/raw/diabetic_data.csv"


df = pd.read_csv(file_path, dtype=str, keep_default_na=False, na_values=["?"])
print(df.head())
```

```python
len(df)
```

```python
n_enc = df['encounter_id'].nunique()
print(n_enc)
```

```python
n_pat = df['patient_nbr'].nunique()
print(n_pat)
```

```python
print(f"Unique encounters: {n_enc:,}  (duplicates: {len(df) - n_enc:,})")
```

```python
enc_per_pat = df.groupby("patient_nbr").size()
```

```python
(enc_per_pat>1).sum()
enc_per_pat[enc_per_pat >1 ].sum()
enc_per_pat.value_counts().sort_index()
```

```python
print(f"Fullly duplicated rows : {df.duplicated().sum(): ,}")
print(f"Max encounters for one patient: {enc_per_pat.max()}")
print(f"Patients with 10+ encounters: {(enc_per_pat >= 10).sum():,}")
```
